```python
def convert(obj):
    # If already a list, return as is
    if isinstance(obj, list):
        return obj

    # If string, convert to list
    if isinstance(obj, str):
        # Remove brackets and quotes
        obj = obj.strip().strip("[]").replace("'", "").replace('"', '')
        
        # Handle empty string
        if not obj:
            return []
        
        # Split by comma and clean spaces
        return [item.strip() for item in obj.split(",")]

    # Unsupported type
    raise ValueError("Input must be a string or a list")
```

```python
def score_to_text(score):
    if score >= 8:
        return "excellent_score"
    elif score >= 6:
        return "good_score"
    elif score >= 4:
        return "average_score"
    else:
        return "low_score"
```

```python
df["score"]=df['score'].apply(score_to_text)
```

```python
#bags of words
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv=CountVectorizer(max_features=2000,stop_words="english")
```

```python
vectors=cv.fit_transform(new_df["tags"]).toarray()
```

```python
vectors.shape
```

```python
cv.get_feature_names_out()
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity=cosine_similarity(vectors)
```

```python
similarity
```

```python
new_df[new_df["name"]=="jujutsu kaisen"]
```

```python
def recommend(anime):
    anime_index = new_df[new_df['tittle'] == anime].index[0]
    distances = similarity[anime_index]
    anime_list = sorted(list(enumerate(distances)), reverse=True, key=lambda x: x[1])[1:6]
    for i in anime_list:
        print(new_df.iloc[i[0]].tittle)
        
```

```python
recommend("jujutsu kaisen")
```
